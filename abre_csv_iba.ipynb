```python
import numpy as np
import matplotlib.pyplot as plt


def load_dose_matrix(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        lines = f.readlines()

    # Extraer eje X (segunda línea)
    x_coords = np.fromstring(lines[1].strip(','), sep=',')

    # Inicializar listas
    y_coords = []
    dose_matrix = []

    # Recorrer las líneas restantes para eje Y y matriz de dosis
    for line in lines[2:]:
        parts = line.strip().split(',')
        y_coords.append(float(parts[0]))
        dose_matrix.append([float(val) for val in parts[1:]])

    y_coords = np.array(y_coords)
    dose_matrix = np.array(dose_matrix)

    return x_coords, y_coords, dose_matrix


# Ejemplo de uso:
file_path = './Data/MX-QAR-VMDmLNmBOOSTVMAT.csv'  # Ruta al archivo CSV
x, y, dose = load_dose_matrix(file_path)

plt.imshow(dose, extent=[x.min(), x.max(), y.min(), y.max()], origin='lower', cmap='turbo', aspect='auto')
plt.colorbar(label='Dose (mGy)')
plt.title('Matriz de Dosis')
plt.xlabel('X (mm)')
plt.ylabel('Y (mm)')
plt.show()

print(dose.shape)
```
